```python
master = master.sort_values(["patient_id", "admission_date"])
```

```python
master["previous_discharge"] = master.groupby("patient_id")["discharge_date"].shift(1)
```

```python
master["days_after_discharge"] = (
    master["admission_date"] - master["previous_discharge"]
).dt.days
```

```python
master["readmission"] = master["days_after_discharge"].apply(
    lambda x: "Yes" if pd.notna(x) and x <= 30 else "No"
)
```
