# KPI Engineering

```python
import pandas as pd

master = pd.read_csv("Processed_Data/Fact_Admissions_Clean.csv")

master.head()
```

```python
doctor = pd.read_csv("Processed_Data/Dim_Doctor.csv")
ward = pd.read_csv("Processed_Data/Dim_Ward.csv")
bed = pd.read_csv("Processed_Data/bed.csv")
```

```python
print(master.shape)

master.info()
```

```python
total_patients = master["patient_id"].nunique()

print("Total Patients:", total_patients)
```

```python
total_admissions = master["admission_id"].nunique()

print("Total Admissions:", total_admissions)
```

```python
master["admission_date"] = pd.to_datetime(master["admission_date"])
master["discharge_date"] = pd.to_datetime(master["discharge_date"])
```

```python
total_wards = master["ward_name"].nunique()

print("Total Wards:", total_wards)
```

```python
average_los = round(master["length_of_stay"].mean(), 2)

print("Average Length of Stay:", average_los)
```

```python
department_admissions = (
    master.groupby("department_name")
    .size()
    .reset_index(name="Total Admissions")
    .sort_values(by="Total Admissions", ascending=False)
)

department_admissions
```

```python
disease_analysis = (
    master.groupby("disease_name")
    .size()
    .reset_index(name="Total Admissions")
    .sort_values(by="Total Admissions", ascending=False)
)

disease_analysis
```

```python
monthly_admissions = (
    master.groupby("admission_month")
    .size()
    .reset_index(name="Total Admissions")
)

monthly_admissions
```

```python
gender_distribution = (
    master.groupby("gender")
    .size()
    .reset_index(name="Patients")
)

gender_distribution
```

```python
admission_type = (
    master.groupby("admission_type")
    .size()
    .reset_index(name="Admissions")
)

admission_type
```

```python
bed_status = (
    master.groupby("bed_status")
    .size()
    .reset_index(name="Beds")
)

bed_status
```

```python
department_admissions.to_csv("KPI_Data/Department_Admissions.csv", index=False)

disease_analysis.to_csv("KPI_Data/Disease_Analysis.csv", index=False)

monthly_admissions.to_csv("KPI_Data/Monthly_Admissions.csv", index=False)

gender_distribution.to_csv("KPI_Data/Gender_Distribution.csv", index=False)

admission_type.to_csv("KPI_Data/Admission_Type.csv", index=False)

bed_status.to_csv("KPI_Data/Bed_Status.csv", index=False)

print(" KPI files saved successfully!")
```

```python
city_patients = (
    master.groupby("city")
    .size()
    .reset_index(name="Patients")
)

city_patients.to_csv("KPI_Data/City_Patients.csv", index=False)
city_patients.head()
```

```python
blood_group = (
    master.groupby("blood_group")
    .size()
    .reset_index(name="Patients")
)

blood_group.to_csv("KPI_Data/Blood_Group.csv", index=False)
blood_group.head()
```

```python
disease_category = (
    master.groupby("disease_category")
    .size()
    .reset_index(name="Patients")
)

disease_category.to_csv("KPI_Data/Disease_Category.csv", index=False)
disease_category.head()
```

```python
master = master.sort_values(["patient_id", "admission_date"])

master["previous_discharge"] = master.groupby("patient_id")["discharge_date"].shift(1)

master["days_after_discharge"] = (
    master["admission_date"] - master["previous_discharge"]
).dt.days

master["readmission"] = master["days_after_discharge"].apply(
    lambda x: "Yes" if pd.notna(x) and x <= 30 else "No"
)
```

```python
readmission_summary = (
    master.groupby("readmission")
          .size()
          .reset_index(name="Total_Patients")
)

readmission_summary
```

```python
total_admissions = len(master)

readmission_rate = (
    master["readmission"].eq("Yes").sum() / total_admissions
) * 100

print(f"Readmission Rate: {readmission_rate:.2f}%")
```

```python
readmission_summary.to_csv(
    "KPI_Data/Readmission_Summary.csv",
    index=False
)

print("✅ Readmission_Summary.csv saved successfully!")
```

```python
# Resource KPIs

total_beds = ward["total_beds"].sum()

total_doctors = doctor["doctor_id"].nunique()

avg_doctor_experience = round(doctor["experience_years"].mean(), 2)

available_beds = bed[bed["bed_status"] == "Available"]["bed_id"].count()

occupied_beds = bed[bed["bed_status"] == "Occupied"]["bed_id"].count()

bed_occupancy_rate = round((occupied_beds / total_beds) * 100, 2)
```

```python
kpi_summary = pd.DataFrame({

    "KPI":[
        "Total Admissions",
        "Total Patients",
        "Total Departments",
        "Total Wards",
        "Average Length of Stay",
        "Readmission Rate (%)",
        "Total Beds",
        "Total Doctors",
        "Average Doctor Experience",
        "Available Beds",
        "Occupied Beds",
        "Bed Occupancy Rate (%)"
    ],

    "Value":[
        total_admissions,
        total_patients,
        master["department_name"].nunique(),
        total_wards,
        average_los,
        round((master["readmission"]=="Yes").mean()*100,2),
        total_beds,
        total_doctors,
        avg_doctor_experience,
        available_beds,
        occupied_beds,
        bed_occupancy_rate
    ]

})
```

```python
kpi_summary.to_csv("KPI_Data/KPI_Summary.csv", index=False)

print("KPI_Summary.csv saved successfully!")
```
