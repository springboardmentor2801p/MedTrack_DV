```python
# ==========================================================
# MedTrack - Module 3
# generate_hospital_kpis.py
# Hospital KPI Engineering
# ==========================================================

import pandas as pd

# ----------------------------------------------------------
# Load Dataset
# ----------------------------------------------------------

file_path = "hospital_final_dataset.xlsx"

df = pd.read_excel(file_path)

print("=" * 60)
print("         MEDTRACK HOSPITAL KPI REPORT")
print("=" * 60)

# ----------------------------------------------------------
# KPI 1 : Total Admissions
# ----------------------------------------------------------

total_admissions = len(df)

# ----------------------------------------------------------
# KPI 2 : Occupancy Rate
# ----------------------------------------------------------

occupancy_rate = df["Occupancy_Rate"].mean()

# ----------------------------------------------------------
# KPI 3 : Average Length of Stay
# ----------------------------------------------------------

average_length_of_stay = df["time_in_hospital"].mean()

# ----------------------------------------------------------
# KPI 4 : Readmission Rate
# (Patients readmitted within 30 days or after 30 days)
# ----------------------------------------------------------

readmission_rate = (
    (df["readmitted"] != "NO").sum() / total_admissions
) * 100

# ----------------------------------------------------------
# KPI 5 : Bed Utilization Rate
# ----------------------------------------------------------

bed_utilization_rate = df["Bed_Utilization_Rate"].mean()

# ----------------------------------------------------------
# KPI 6 : Department Efficiency Score
# ----------------------------------------------------------

department_efficiency_score = df["Department_Efficiency_Score"].mean()

# ==========================================================
# Display KPI Report
# ==========================================================

print(f"\n1. Total Admissions              : {total_admissions}")

print(f"2. Occupancy Rate                : {occupancy_rate:.2f}%")

print(f"3. Average Length of Stay        : {average_length_of_stay:.2f} Days")

print(f"4. Readmission Rate              : {readmission_rate:.2f}%")

print(f"5. Bed Utilization Rate          : {bed_utilization_rate:.2f}%")

print(f"6. Department Efficiency Score   : {department_efficiency_score:.2f}")

print("\n" + "=" * 60)
print("Module 3 KPI Calculation Completed Successfully!")
print("=" * 60)
```
