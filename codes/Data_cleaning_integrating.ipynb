```python
import pandas as pd
```

```python
services_weekly=pd.read_csv(r"D:\infosys_Springboard\finalized datasets\services_weekly.csv")
```

```python
patients=pd.read_csv(r"D:\infosys_Springboard\finalized datasets\patients.csv")
```

```python
patients.head()
```

```python
patients["arrival_date"] = pd.to_datetime(patients["arrival_date"])
patients["departure_date"] = pd.to_datetime(patients["departure_date"])
```

```python
patients["week"] = patients["arrival_date"].dt.isocalendar().week.astype(int)
```

```python
patients.columns
```

```python
patients["length_of_stay"] = (
    patients["departure_date"] - patients["arrival_date"]
).dt.days
```

```python
patient_summary = (
    patients
    .groupby(["week", "service"])
    .agg(
        patient_count=("patient_id", "count"),
        average_length_of_stay=("length_of_stay", "mean"),
        average_satisfaction=("satisfaction", "mean")
    )
    .reset_index()
)
```

```python
integrated = pd.merge(
    services_weekly,
    patient_summary,
    on=["week", "service"],
    how="left"
)
```

```python
staff_schedule=pd.read_csv(r"D:\infosys_Springboard\finalized datasets\staff_schedule.csv")
```

```python
staff_summary = (
    staff_schedule
    .groupby(["week", "service"])
    .agg(
        staff_present=("present", "sum")
    )
    .reset_index()
)
```

```python
staff_schedule.head()
```

```python
integrated = integrated.merge(
    staff_summary,
    on=["week", "service"],
    how="left"
)
```

```python
staff=pd.read_csv(r"D:\infosys_Springboard\finalized datasets\staff.csv")
```

```python
staff_master = (
    staff
    .groupby("service")
    .agg(
        total_staff=("staff_id", "count")
    )
    .reset_index()
)
```

```python
integrated = integrated.merge(
    staff_master,
    on="service",
    how="left"
)
```

```python
integrated.to_csv(
    "hospital_integrated_dataset.csv",
    index=False
)
```
