```python
doctor_distribution = {
    101: 20,   # Cardiology
    102: 15,   # Neurology
    103: 18,   # Orthopedics
    104: 25,   # Emergency
    105: 40,   # General Medicine
    106: 22,   # Surgery
    107: 10    # Oncology
}
```

```python
doctors = []

doctor_number = 1

for _, dept in df_departments.iterrows():

    dept_id = dept["Department_ID"]
    dept_name = dept["Department_Name"]

    for i in range(doctor_distribution[dept_id]):

        gender = random.choice(["Male", "Female"])

        if gender == "Male":
            first = fake.first_name_male()
        else:
            first = fake.first_name_female()

        last = fake.last_name()

        experience = random.randint(2, 30)

        salary = random.randint(80000, 300000)

        doctors.append({

            "Doctor_ID": f"D{doctor_number:04}",

            "Doctor_First_Name": first,

            "Doctor_Last_Name": last,

            "Gender": gender,

            "Department_ID": dept_id,

            "Department_Name": dept_name,

            "Qualification": qualification_map[dept_name],

            "Experience_Years": experience,

            "Salary": salary,

            "Phone": fake.phone_number(),

            "Email": f"{first.lower()}.{last.lower()}@hospital.com"

        })

        doctor_number += 1
```

```python
blood_groups = [
    "A+","A-","B+","B-",
    "AB+","AB-","O+","O-"
]

cities = [
    "Hyderabad",
    "Bengaluru",
    "Chennai",
    "Mumbai",
    "Delhi",
    "Pune",
    "Vijayawada",
    "Visakhapatnam"
]

state_map = {
    "Hyderabad":"Telangana",
    "Bengaluru":"Karnataka",
    "Chennai":"Tamil Nadu",
    "Mumbai":"Maharashtra",
    "Delhi":"Delhi",
    "Pune":"Maharashtra",
    "Vijayawada":"Andhra Pradesh",
    "Visakhapatnam":"Andhra Pradesh"
}

insurance_companies = [
    "Star Health",
    "Apollo Munich",
    "ICICI Lombard",
    "HDFC ERGO",
    "Niva Bupa",
    "Care Health"
]
```

```python
patients = []

for i in range(1, NUM_PATIENTS + 1):

    gender = random.choice(["Male","Female"])

    if gender == "Male":
        first = fake.first_name_male()
    else:
        first = fake.first_name_female()

    last = fake.last_name()

    dob = fake.date_of_birth(
        minimum_age=18,
        maximum_age=90
    )

    city = random.choice(cities)

    insured = random.choice(["Yes","No"])

    provider = (
        random.choice(insurance_companies)
        if insured == "Yes"
        else "None"
    )

    patients.append({

        "Patient_ID": f"P{i:05}",

        "First_Name": first,

        "Last_Name": last,

        "Gender": gender,

        "DOB": dob,

        "Blood_Group": random.choice(blood_groups),

        "City": city,

        "State": state_map[city],

        "Phone": fake.phone_number(),

        "Insurance": insured,

        "Insurance_Provider": provider

    })
```

```python
dept_map = (
    df_departments
    .set_index("Department_Name")["Department_ID"]
    .to_dict()
)

dept_map
```

```python
diagnosis_rules = {
    "Heart Disease": {
        "Department": "Cardiology",
        "Stay": (5, 12),
        "Ward": ["ICU", "Private"]
    },

    "Hypertension": {
        "Department": "Cardiology",
        "Stay": (2, 5),
        "Ward": ["General", "Private"]
    },

    "Stroke": {
        "Department": "Neurology",
        "Stay": (7, 15),
        "Ward": ["ICU", "Private"]
    },

    "Migraine": {
        "Department": "Neurology",
        "Stay": (1, 3),
        "Ward": ["General"]
    },

    "Fracture": {
        "Department": "Orthopedics",
        "Stay": (5, 12),
        "Ward": ["Semi-Private", "Private"]
    },

    "Appendicitis": {
        "Department": "Surgery",
        "Stay": (3, 7),
        "Ward": ["General", "Private"]
    },

    "Cancer": {
        "Department": "Oncology",
        "Stay": (10, 30),
        "Ward": ["Private"]
    },

    "Pneumonia": {
        "Department": "General Medicine",
        "Stay": (4, 8),
        "Ward": ["General", "Semi-Private"]
    },

    "Asthma": {
        "Department": "General Medicine",
        "Stay": (2, 5),
        "Ward": ["General"]
    },

    "Diabetes": {
        "Department": "General Medicine",
        "Stay": (2, 4),
        "Ward": ["General"]
    },

    "Kidney Stone": {
        "Department": "General Medicine",
        "Stay": (3, 6),
        "Ward": ["Semi-Private", "Private"]
    },

    "COVID-19": {
        "Department": "Emergency",
        "Stay": (5, 10),
        "Ward": ["ICU", "General"]
    }
}
```

```python
len(diagnosis_rules)
```

```python
import random
from datetime import timedelta

admissions = []

for i in range(1, 10001):

    # Select random patient
    patient = df_patients.sample(1).iloc[0]

    # Select diagnosis
    diagnosis = random.choice(list(diagnosis_rules.keys()))
    rule = diagnosis_rules[diagnosis]

    # Department
    department_name = rule["Department"]
    department_id = dept_map[department_name]

    # Doctor ONLY from that department
    doctor = df_doctors[
        df_doctors["Department_ID"] == department_id
    ].sample(1).iloc[0]

    # Admission date
    admission_date = fake.date_between(
        start_date="-2y",
        end_date="today"
    )

    # Length of stay
    stay = random.randint(
        rule["Stay"][0],
        rule["Stay"][1]
    )

    discharge_date = admission_date + timedelta(days=stay)

    # Ward
    ward = random.choice(rule["Ward"])

    # Room Number
    if ward == "ICU":
        room = f"ICU-{random.randint(1,20)}"
    elif ward == "Private":
        room = f"P-{random.randint(101,999)}"
    elif ward == "Semi-Private":
        room = f"SP-{random.randint(101,999)}"
    else:
        room = f"G-{random.randint(101,999)}"

    admissions.append({

        "Admission_ID": f"A{i:06d}",
        "Patient_ID": patient["Patient_ID"],
        "Doctor_ID": doctor["Doctor_ID"],
        "Department_ID": department_id,

        "Admission_Date": admission_date,
        "Discharge_Date": discharge_date,

        "Admission_Type": random.choice(
            ["Emergency", "Referral", "Elective"]
        ),

        "Diagnosis": diagnosis,
        "Ward_Type": ward,
        "Room_No": room,
        "Length_of_Stay": stay,
        "Status": "Discharged"

    })
```

```python
billing = []

for _, row in df_admissions.iterrows():

    consultation = random.randint(500, 2000)

    room_charge = row["Length_of_Stay"] * random.randint(2000, 6000)

    medicine = random.randint(1000, 15000)

    lab = random.randint(500, 10000)

    other = random.randint(0, 5000)

    total = (
        consultation
        + room_charge
        + medicine
        + lab
        + other
    )

    insurance = random.choice(["Yes", "No"])

    if insurance == "Yes":
        insurance_amount = round(total * random.uniform(0.5, 0.9), 2)
    else:
        insurance_amount = 0

    patient_pay = round(total - insurance_amount, 2)

    billing.append({

        "Bill_ID": f"B{len(billing)+1:06d}",

        "Admission_ID": row["Admission_ID"],

        "Consultation_Fee": consultation,

        "Room_Charges": room_charge,

        "Medicine_Cost": medicine,

        "Lab_Cost": lab,

        "Other_Charges": other,

        "Total_Amount": total,

        "Insurance_Covered": insurance_amount,

        "Patient_Payable": patient_pay,

        "Payment_Status": random.choice([
            "Paid",
            "Pending",
            "Partial"
        ])
    })
```
