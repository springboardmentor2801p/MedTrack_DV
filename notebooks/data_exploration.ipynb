```python
department_summary = (
    df.groupby('department_name')
    .agg(
        total_admissions=('admission_id','count'),
        total_wards=('ward_name','nunique'),
        total_diseases=('disease_name','nunique')
    )
    .reset_index()
)

department_summary
```

```python
department_summary.to_csv(
    "../data/processed/department_summary.csv",
    index=False
)
```

```python
disease_summary = (
    df.groupby(['disease_category'])
    .agg(
        total_cases=('admission_id','count'),
        unique_diseases=('disease_name','nunique')
    )
    .reset_index()
)

disease_summary
```

```python
disease_summary = (
    df.groupby(['disease_category'])
    .agg(
        total_cases=('admission_id','count'),
        unique_diseases=('disease_name','nunique')
    )
    .reset_index()
)

disease_summary
```

```python
disease_summary.to_csv(
    "../data/processed/disease_summary.csv",
    index=False
)
```
