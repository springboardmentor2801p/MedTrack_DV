```python
import pandas as pd
import numpy as np
```

```python
path = "../data/processed/hospital_raw_data.csv"

df = pd.read_csv(path)

print(df.shape)
df.head()
```

```python
missing = df.isnull().sum()

missing[missing > 0]
```

```python
print("Before:", df.shape)

df = df.drop_duplicates()

print("After:", df.shape)
```

```python
patient_cols = [
    "gender",
    "date_of_birth",
    "blood_group",
    "city",
    "contact_number"
]

for col in patient_cols:
    df[col] = df[col].fillna("Unknown")
```

```python
output = "../data/cleaned/hospital_cleaned.csv"

df.to_csv(output,index=False)

print("Clean file created")
```

```python
missing_percent = (df.isnull().sum() / len(df)) * 100

missing_percent[
    missing_percent > 0
]
```
