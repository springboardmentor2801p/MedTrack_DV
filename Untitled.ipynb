```python
import pandas as pd
import numpy as np

patients = pd.read_csv("diabetic_data.csv")
hospital = pd.read_csv("HospInfo.csv")
provider = pd.read_csv("pos2011.csv")
mapping = pd.read_csv("IDS_mapping.csv")
```

```python
patients = patients.drop_duplicates()
patients.replace("?", np.nan, inplace=True)
```

```python
patients = patients.sample(n=5000, random_state=42).reset_index(drop=True)
```

```python
hospital = hospital[
[
'Provider ID',
'Hospital Name',
'City',
'State',
'Hospital Type',
'Hospital Ownership',
'Emergency Services',
'Hospital overall rating'
]]
```

```python
provider = provider[
[
'PRVDR_NUM',
'FAC_NAME',
'CITY_NAME',
'STATE_CD',
'BED_CNT',
'EMPLEE_CNT'
]]
```

```python
hospital = hospital.head(5000).reset_index(drop=True)
provider = provider.head(5000).reset_index(drop=True)
```

```python
provider.rename(columns={
    'FAC_NAME':'Facility Name',
    'CITY_NAME':'Provider City',
    'STATE_CD':'Provider State'
}, inplace=True)
```

```python
patients["temp_id"] = patients.index
hospital["temp_id"] = hospital.index
provider["temp_id"] = provider.index
```

```python
final = patients.merge(
    hospital,
    on="temp_id",
    how="left"
)

final = final.merge(
    provider,
    on="temp_id",
    how="left"
)
```

```python
final.drop("temp_id", axis=1, inplace=True)
```

```python
final.to_csv("hospital_cleaned.csv", index=False)
```
